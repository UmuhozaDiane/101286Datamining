```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from ISLP import load_data
from sklearn.model_selection import train_test_split
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis, QuadraticDiscriminantAnalysis
from sklearn.metrics import accuracy_score, confusion_matrix
```

```python
print({"Fraction_used_p1": 0.10, "Fraction_used_p2": 0.10 ** 2, "Fraction_used_p3": 0.10 ** 3, "Fraction_used_p10": 0.10 ** 10})
```

```python
odds = 0.37
probability = odds / (1 + odds)
p = 0.16
odds_from_probability = p / (1 - p)
print({"Probability_from_odds_0.37": probability, "Odds_from_probability_0.16": odds_from_probability})
```

```python
Auto = load_data("Auto").dropna()
Auto["mpg01"] = (Auto["mpg"] > Auto["mpg"].median()).astype(int)
numeric_features = ["cylinders", "displacement", "horsepower", "weight", "acceleration", "year"]
print(Auto.groupby("mpg01")[numeric_features].mean())
for feature in numeric_features:
    plt.figure(figsize=(6, 4))
    Auto.boxplot(column=feature, by="mpg01")
    plt.title(feature + " by mpg01")
    plt.suptitle("")
    plt.tight_layout()
    plt.show()
```

```python
features = ["cylinders", "displacement", "horsepower", "weight", "year"]
X = Auto[features]
y = Auto["mpg01"]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=1, stratify=y)
for name, estimator in [("LDA", LinearDiscriminantAnalysis()), ("QDA", QuadraticDiscriminantAnalysis())]:
    estimator.fit(X_train, y_train)
    prediction = estimator.predict(X_test)
    print(name, {"Test_Error": 1 - accuracy_score(y_test, prediction), "Confusion_Matrix": confusion_matrix(y_test, prediction).tolist()})
```
