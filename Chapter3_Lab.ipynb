```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import statsmodels.api as sm
from ISLP import load_data
from ISLP.models import ModelSpec as MS
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
```

```python
Auto = load_data("Auto").dropna()
print(Auto.head())
print(Auto.describe())
```

```python
plt.figure(figsize=(8, 5))
plt.scatter(Auto["horsepower"], Auto["mpg"])
plt.xlabel("Horsepower")
plt.ylabel("MPG")
plt.tight_layout()
plt.show()
```

```python
X = sm.add_constant(Auto[["horsepower"]])
model = sm.OLS(Auto["mpg"], X).fit()
print(model.summary())
```

```python
new_data = pd.DataFrame({"const": [1, 98], "horsepower": [98, 98]})
print(model.predict(new_data))
```

```python
plt.figure(figsize=(8, 5))
plt.scatter(Auto["horsepower"], Auto["mpg"], alpha=0.6)
order = np.argsort(Auto["horsepower"])
plt.plot(Auto["horsepower"].iloc[order], model.predict(X).iloc[order])
plt.xlabel("Horsepower")
plt.ylabel("MPG")
plt.tight_layout()
plt.show()
```

```python
predictors = ["horsepower", "weight", "displacement", "acceleration"]
X = sm.add_constant(Auto[predictors])
multiple_model = sm.OLS(Auto["mpg"], X).fit()
print(multiple_model.summary())
```

```python
X_train, X_test, y_train, y_test = train_test_split(Auto[predictors], Auto["mpg"], test_size=0.3, random_state=1)
reg = LinearRegression().fit(X_train, y_train)
pred = reg.predict(X_test)
print({"Test_MSE": mean_squared_error(y_test, pred), "R_squared": reg.score(X_test, y_test)})
```
