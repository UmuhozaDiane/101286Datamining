```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from ISLP import load_data
from sklearn.model_selection import train_test_split, cross_val_score, KFold
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

```python
n_values = np.arange(1, 100001)
probability = 1 - (1 - 1 / n_values) ** n_values
print(pd.DataFrame({"n": [5, 100, 10000], "Probability_in_bootstrap": 1 - (1 - 1 / np.array([5, 100, 10000])) ** np.array([5, 100, 10000])}))
plt.figure(figsize=(8, 5))
plt.plot(n_values, probability)
plt.xlabel("Sample size n")
plt.ylabel("Probability observation appears in bootstrap sample")
plt.tight_layout()
plt.show()
```

```python
Default = load_data("Default")
X = Default[["income", "balance"]]
y = (Default["default"] == "Yes").astype(int)
errors = []
for seed in [1, 2, 3, 4]:
    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.5, random_state=seed, stratify=y)
    model = LogisticRegression(max_iter=2000).fit(X_train, y_train)
    errors.append({"Seed": seed, "Validation_Error": 1 - accuracy_score(y_test, model.predict(X_test))})
print(pd.DataFrame(errors))
```

```python
cv = KFold(n_splits=10, shuffle=True, random_state=1)
cv_errors = 1 - cross_val_score(LogisticRegression(max_iter=2000), X, y, cv=cv, scoring="accuracy")
print({"10_fold_CV_Error": cv_errors.mean(), "Fold_Errors": cv_errors})
```
