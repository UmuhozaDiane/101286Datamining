```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from ISLP import load_data
from sklearn.model_selection import train_test_split, GridSearchCV, StratifiedKFold
from sklearn.svm import SVC
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.metrics import accuracy_score, confusion_matrix, RocCurveDisplay, roc_auc_score
```

```python
rng = np.random.default_rng(1)
X = rng.standard_normal((50, 2))
y = np.array([-1] * 25 + [1] * 25)
X[y == 1] += 1
plt.figure(figsize=(7, 5))
plt.scatter(X[:, 0], X[:, 1], c=y)
plt.tight_layout()
plt.show()
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=1, stratify=y)
for kernel in ["linear", "rbf", "poly"]:
    model = make_pipeline(StandardScaler(), SVC(C=1, kernel=kernel, probability=True, random_state=1))
    model.fit(X_train, y_train)
    pred = model.predict(X_test)
    print(kernel, {"Accuracy": accuracy_score(y_test, pred), "Confusion_Matrix": confusion_matrix(y_test, pred).tolist()})
```

```python
Smarket = load_data("Smarket")
X = Smarket[["Lag1", "Lag2"]]
y = Smarket["Direction"]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=1, stratify=y)
param_grid = {"svc__C": [0.01, 0.1, 1, 10], "svc__kernel": ["linear", "rbf"]}
model = make_pipeline(StandardScaler(), SVC())
grid = GridSearchCV(model, param_grid, cv=StratifiedKFold(n_splits=5, shuffle=True, random_state=1), scoring="accuracy")
grid.fit(X_train, y_train)
pred = grid.predict(X_test)
print({"Best_Parameters": grid.best_params_, "CV_Accuracy": grid.best_score_, "Test_Accuracy": accuracy_score(y_test, pred)})
print(confusion_matrix(y_test, pred))
```
