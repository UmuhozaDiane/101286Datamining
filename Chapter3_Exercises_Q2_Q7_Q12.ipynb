```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import statsmodels.api as sm
from ISLP import load_data
```

```python
Advertising = load_data("Advertising")
print(Advertising.head())
X = sm.add_constant(Advertising[["TV", "radio", "newspaper"]])
model = sm.OLS(Advertising["sales"], X).fit()
print(model.summary())
```

```python
Auto = load_data("Auto").dropna()
simple = sm.OLS(Auto["mpg"], sm.add_constant(Auto["horsepower"])).fit()
print(simple.summary())
print({"Slope": simple.params["horsepower"], "Slope_p_value": simple.pvalues["horsepower"], "R_squared": simple.rsquared})
```

```python
rng = np.random.default_rng(1)
n = 100
p = 100
X = rng.normal(size=(n, p))
beta = rng.normal(size=p)
Y = 1 + X @ beta + rng.normal(size=n)
beta_multiple = np.linalg.lstsq(np.column_stack([np.ones(n), X]), Y, rcond=None)[0]
residual = Y.copy()
beta_backfit = np.zeros(p)
for iteration in range(1, 1001):
    old = beta_backfit.copy()
    for j in range(p):
        residual = Y - (beta_backfit @ X.T) + beta_backfit[j] * X[:, j]
        beta_backfit[j] = X[:, j] @ (residual - residual.mean()) / (X[:, j] @ X[:, j])
    if np.max(np.abs(beta_backfit - old)) < 1e-6:
        break
beta0_backfit = np.mean(Y - X @ beta_backfit)
print({"Backfitting_iterations": iteration, "Intercept": beta0_backfit, "Max_abs_slope_difference": np.max(np.abs(beta_backfit - beta_multiple[1:]))})
```

```python
plt.figure(figsize=(9, 5))
plt.plot(beta_multiple[1:], label="Multiple regression coefficients")
plt.plot(beta_backfit, label="Backfitting coefficients")
plt.xlabel("Predictor")
plt.ylabel("Coefficient")
plt.legend()
plt.tight_layout()
plt.show()
```
