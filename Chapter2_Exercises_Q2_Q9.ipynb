```python
from ISLP import load_data
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsRegressor
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.metrics import mean_squared_error
```

```python
scenarios = pd.DataFrame([
    {"Scenario": "2(a)", "Problem": "Regression", "Goal": "Inference", "n": 500, "p": 3},
    {"Scenario": "2(b)", "Problem": "Classification", "Goal": "Prediction", "n": 20, "p": 13},
    {"Scenario": "2(c)", "Problem": "Regression", "Goal": "Prediction", "n": 52, "p": 3}
])
print(scenarios)
```

```python
Auto = load_data("Auto")
print(Auto.head())
print(Auto.describe(include="all"))
print(Auto.isna().sum())
```

```python
Auto = Auto.dropna()
X = Auto[["horsepower"]]
y = Auto["mpg"]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=1)
results = []
for k in range(1, 21):
    model = make_pipeline(StandardScaler(), KNeighborsRegressor(n_neighbors=k))
    model.fit(X_train, y_train)
    predictions = model.predict(X_test)
    results.append({"K": k, "Test_MSE": mean_squared_error(y_test, predictions)})
results = pd.DataFrame(results)
print(results)
print(results.loc[results["Test_MSE"].idxmin()])
```

```python
plt.figure(figsize=(8, 5))
plt.scatter(Auto["horsepower"], Auto["mpg"], alpha=0.7)
plt.xlabel("Horsepower")
plt.ylabel("Miles per gallon")
plt.title("Auto: MPG versus Horsepower")
plt.tight_layout()
plt.show()
```
